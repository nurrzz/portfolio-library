# Netflix Catalog Analysis | Movies vs. TV Shows

I wanted to understand the content mix in a snapshot of Netflix’s catalog. A content team could use this breakdown as a starting point before comparing the catalog across release years or markets.

**Research question:** What share of the titles in this dataset are movies, and what share are TV shows?

**Data:** I use the [Netflix Movies and TV Shows dataset on Kaggle](https://www.kaggle.com/datasets/shivamb/netflix-shows). Each row represents one listing. I use `show_id` and `title` to preview the records and `type` to count movies and TV shows.

## 1. Download the dataset

I use KaggleHub to download the latest version of [Netflix Movies and TV Shows](https://www.kaggle.com/datasets/shivamb/netflix-shows). Run the cells in order. The first cell installs KaggleHub in Colab.

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("shivamb/netflix-shows")

print("Path to dataset files:", path)
```

## 2. Preview the data

I inspect five rows to see what the listing ID, title, and content type look like before counting the full dataset.

```python
import csv
from itertools import islice
from pathlib import Path

csv_path = Path(path) / "netflix_titles.csv"

with csv_path.open(encoding="utf-8-sig", newline="") as csv_file:
    titles = csv.DictReader(csv_file)
    for row in islice(titles, 5):
        print(row["show_id"], row["title"], row["type"], sep=" | ")
```

## 3. Measure the content mix

I count each movie and TV show listing once. For each category, I divide its count by the combined movie and TV show count. A series counts as one listing regardless of its number of seasons.

```python
movie_count = 0
tv_show_count = 0

try:
    with csv_path.open(encoding="utf-8-sig", newline="") as csv_file:
        titles = csv.DictReader(csv_file)
        if "type" not in (titles.fieldnames or []):
            raise SystemExit("The CSV needs a column named 'type'.")

        for title in titles:
            if title["type"] == "Movie":
                movie_count += 1
            elif title["type"] == "TV Show":
                tv_show_count += 1
except FileNotFoundError:
    raise SystemExit("Could not find netflix_titles.csv in the Kaggle download folder.")

total = movie_count + tv_show_count
if total == 0:
    raise SystemExit("No movies or TV shows were found in the CSV.")

print(f"Movies:   {movie_count:,} ({movie_count / total:.1%})")
print(f"TV shows: {tv_show_count:,} ({tv_show_count / total:.1%})")

if movie_count > tv_show_count:
    print("Movies make up the larger share of this catalog snapshot.")
elif tv_show_count > movie_count:
    print("TV shows make up the larger share of this catalog snapshot.")
else:
    print("Movies and TV shows make up equal shares of this catalog snapshot.")
```

## 4. What the result tells me

The percentages describe the balance of listings in this dataset. Viewing behavior and subscriber preferences would require separate data. A follow-up using this CSV could compare the movie-to-TV mix by release decade to see how the snapshot represents older and newer titles.
